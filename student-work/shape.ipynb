```python
class Shape:
    def __init__(self, name):
        self.name=name
    
    def area(self):
        print("area of shape")       
```

```python
class Circle(Shape):
    def __init__(self, name, radius):
        super().__init__(name)
        self.radius=radius
    def area(self):
        return 3.14*self.radius**2
    
```

```python
circle=Circle("c1",5)
circle.area()
```
